```python
import pandas as pd

df = pd.read_csv(r"/content/Day11_Messy_Company_Employee_Dataset.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
#Inspect the dataset

print("Dataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
#Check missing values
missing_values = df.isnull().sum()


print("Missing values in each column:")
print(missing_values)


print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
#Check duplicate records
print("Number of duplicate rows:", df.duplicated().sum())
print("\nDuplicate records:")
display(df[df.duplicated(keep=False)])
```

```python
#Check inconsistent categorical values
categorical_columns = [
    "Department",
    "Gender",
    "City",
    "Work_Mode"
]


for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))
```

```python
#Remove duplicate records
df = df.drop_duplicates()


print("Shape after removing duplicates:", df.shape)
print("Remaining duplicates:", df.duplicated().sum())
```

```python
#Standardize inconsistent categorical values
df["Department"] = df["Department"].astype("string").str.strip().str.title()
df["Gender"] = df["Gender"].astype("string").str.strip().str.title()
df["City"] = df["City"].astype("string").str.strip().str.title()
df["Work_Mode"] = df["Work_Mode"].astype("string").str.strip().str.title()
for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))
```

```python
#Correct data types
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce"
)

#Convert numerical columns:

numeric_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.dtypes)
```

```python
#Fill missing numerical values
numerical_missing_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]
for col in numerical_missing_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
print(df[numerical_missing_columns].isnull().sum())
```

```python
#fill missing categorical values
categorical_missing_columns = [
    "Department",
    "Gender",
    "City",
    "Work_Mode"
]

for col in categorical_missing_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df[categorical_missing_columns].isnull().sum())
```

```python
#Handle essential columns
df = df.dropna(
    subset=["Employee_ID", "Employee_Name", "Job_Title"]
)
```

```python
#Verify the cleaned dataset
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)


display(df.head())
```

```python
#Compare before and after cleaning

# Original dataset statistics
original_df = pd.read_csv(r"/content/Day11_Messy_Company_Employee_Dataset.csv")


before_rows = original_df.shape[0]
before_columns = original_df.shape[1]
before_missing = original_df.isnull().sum().sum()
before_duplicates = original_df.duplicated().sum()


after_rows = df.shape[0]
after_columns = df.shape[1]
after_missing = df.isnull().sum().sum()
after_duplicates = df.duplicated().sum()


comparison = pd.DataFrame({
    "Condition": ["Before Cleaning", "After Cleaning"],
    "Rows": [before_rows, after_rows],
    "Columns": [before_columns, after_columns],
    "Missing Values": [before_missing, after_missing],
    "Duplicate Rows": [before_duplicates, after_duplicates]
})
display(comparison)
```

```python
#Export the cleaned dataset
output_file = "Cleaned_Company_Employee_Dataset.csv"
df.to_csv(output_file, index=False)
print("Cleaned dataset exported successfully!")
```
